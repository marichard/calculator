# Basic Calculator

```python
from pycparser.c_ast import While


# define basic calculator functions
def add(x,y):
    return x+y

def subtract(x,y):
    return x-y

def multiply(x,y):
    return x*y

def divide(x,y):
    return x/y
```

```python
# instructions for user
print("Enter A for addition.")
print("Enter B for subtraction.")
print("Enter C for multiplication.")
print("Enter D for division.")

while True:
    choice = input('Enter choice (A,S,M,D): ')

    if choice.upper() in ('A','S','M','D'):
        num1 = float(input('Enter first number: '))
        num2 = float(input('Enter second number: '))


        if choice.upper() == 'A':
            print('Result:',num1, '+', num2, '=',add(num1,num2))

        elif choice.upper() == 'S':
            print('Result:',num1, '-', num2, '=',subtract(num1,num2))

        elif choice.upper() == 'M':
            print('Result:',num1, '*', num2, '=',multiply(num1,num2))

        elif choice.upper() == 'D':
            print('Result:',num1, '/', num2, '=',divide(num1,num2))

    else:
        print('Please input a correct choice')

    nextCalculation = input("Want to do another calculation? (Y/N): ")
    if nextCalculation.lower() == 'N':
        break
```
